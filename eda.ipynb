```python
train_df['Date'] = pd.to_datetime(train_df["Date"])
train_df.dtypes
```

```python
train_df['year'] = train_df['Date'].dt.year 
train_df['month'] = train_df['Date'].dt.month
train_df['day'] = train_df['Date'].dt.day
train_df['week'] = train_df['Date'].dt.isocalendar().week.astype(int)
train_df.drop(columns= 'Date', inplace=True)
train_df.head()
```

```python
train_df.dtypes
```

```python
target = ["Sales"]
num_cols = ['Customers', 'day', 'month', 'year', 'week']
cat_cols = ['Store', 'DayOfWeek', 'Open', 'Promo', 'StateHoliday', 'SchoolHoliday']
```

```python
train_df[num_cols].describe()
```

univariate analayze

```python
plt.figure(figsize=(6,4))
plt.hist(train_df[target])
plt.show()
```

```python
for col in num_cols :
    plt.figure(figsize=(6,4))
    plt.hist(train_df[col])
    plt.title(col)
    plt.xlabel(col)
    plt.show()
```

bivariate analyze

```python
for col in num_cols :
    plt.Figure(figsize=(6,4))
    sns.regplot(data=train_df, x=col, y=target)
    plt.show()
```

```python
for col in cat_cols :
    plt.figure(figsize=(6,4))
    sns.boxplot(data=train_df, x=col, y=target)
    plt.show()
```
